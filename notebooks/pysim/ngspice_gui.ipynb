````python
import os
import sys
from pathlib import Path
import subprocess
import shlex

# subprocess.call([sys.executable, '-m', 'pip', 'install', 'pandas'])
# cmd = "ls -al | wc -l"
def run_shell(cmd: str, *args, **kwargs):
    proc = subprocess.Popen(cmd, shell=True,
                            stdout = subprocess.PIPE,
                            stderr = subprocess.PIPE,
                            encoding='utf-8'
                            )
    stdout, stderr = proc.communicate()
    if proc.returncode != 0:
        print(stderr)
    else:
        print(stdout)

cmd = "echo Hello!"
run_shell(cmd)
````

````python
from panel.viewable import Viewer

import panel as pn
import param

pn.extension('codeeditor', design='material', theme='default')

class VbCodeEditor(Viewer):

    # editor = param.String('// Code Editor')
    editor = param.String('')

    def __init__(self, language: str = 'python',
                 **params):
        super().__init__(**params)
        self.language = language

    @property
    def filetypes(self):
        _filetype = {
            'python': 'py',
            'spice': 'cir',
            'html': 'html',
            'c': 'c',
            'cpp': 'cpp',
            'java': 'java',
            'javascript': 'js',
            'json': 'json',
        }
        return _filetype

    def __panel__(self):
        """ Map the string to appear as an Ace editor. """
        _filetype = self.filetypes.get(self.language, 'txt')
        return pn.Param(
            self.param,
            widgets=dict(
                editor=dict(
                    type=pn.widgets.CodeEditor,
                    language=self.language,
                    theme = "monokai",
                    # sizing_mode='stretch_width', language='python',
                    height=300,
                    annotations = [
                        dict(row=1, column=0, text='an error', type='error'),
                        dict(row=2, column=0, text='a warning', type='warning')
                    ],
                    filename = f'main.{_filetype}',
                    on_keyup=False
                )
            )
        )

# editor.annotations = [
#     dict(row=1, column=0, text='an error', type='error'),
#     dict(row=2, column=0, text='a warning', type='warning')
# ]

# editor.filename = 'test.html'


edit = VbCodeEditor(language='python')

title_pn = pn.panel('# PySim')
pn.Column(title_pn, edit).servable()

# edit
````

### Code Editor

````python
import panel as pn
import param

pn.extension('codeeditor')



def get_code_editor(language: str = 'python',
                    filename: str | None = None,
                **params):
    editor = pn.widgets.CodeEditor(value='', sizing_mode='stretch_width', language='python', height=300, on_keyup=False)

    editor.language = language
    editor.theme = "monokai"
    editor.value = ""

    editor.annotations = [
        dict(row=1, column=0, text='an error', type='error'),
        dict(row=2, column=0, text='a warning', type='warning')
    ]

    # editor.filename = 'test.py'
    if filename is not None:
        editor.filename = filename


    # editor = pn.widgets.CodeEditor(value=py_code, on_keyup=False)
    return editor

def print_code(value):
    print(value)

def save_code(value, filename = None):
    with open(filename, 'w') as f:
        f.write(value)

def run_code(value):
    print(value)

editor = get_code_editor()
pn.Column(editor,
          pn.bind(run_code, editor.param.value))
````
